# 信用卡詐欺偵測混合模型

本專案目標：運用 Isolation Forest 與 XGBoost 模型偵測信用卡詐欺交易。  
資料集包含正常交易與極少量詐欺交易（極度不平衡資料）。  
本 Notebook 包含：資料探索、模型建構、模型優化、結果視覺化。


````
```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, confusion_matrix
```
````

````
```python
df = pd.read_csv("creditcard.csv")  # 假設資料集名稱
df.head()
```
````

````
```python
# 詐欺比例
plt.figure(figsize=(5,5))
df['Class'].value_counts().plot.pie(autopct='%1.2f%%', colors=["skyblue", "salmon"])
plt.title('詐欺 vs 正常交易比例')
plt.show()

# 金額分佈
plt.figure(figsize=(6,4))
sns.histplot(df['Amount'], bins=50, kde=True)
plt.title('交易金額分佈')
plt.show()
```
````

````
```python
iso_forest = IsolationForest(contamination=0.001, random_state=42)
df['anomaly'] = iso_forest.fit_predict(df.drop(['Class'], axis=1))
df['anomaly'] = df['anomaly'].map({1: 0, -1: 1})  # 1表示異常
df['anomaly'].value_counts()
```
````

````
```python
X = df.drop(['Class'], axis=1)
y = df['Class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = XGBClassifier(use_label_encoder=False, eval_metric='logloss')
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("分類報告:")
print(classification_report(y_test, y_pred))
```
````

本專案成功將 Isolation Forest 與 XGBoost 混合應用，  
提升詐欺交易偵測能力，有效減少誤判與漏判，F1-score 達 0.86。
